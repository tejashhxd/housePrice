```python
x = df.drop(['SalePrice'], axis=1)
```

```python
y = df['SalePrice']
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
X_train, X_test, y_train, y_test = train_test_split(
    x, y, test_size=0.33, random_state=42)
```

```python
X_train_scaled = scaler.fit_transform(X_train)
```

```python
LR_model = model_1.fit(X_train_scaled, y_train)
```

```python
X_test_scaled = scaler.fit_transform(X_test)
```

```python
LR_model.score(X_test_scaled, y_test)
```

```python
from sklearn.metrics import accuracy_score, f1_score, classification_report
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeRegressor
from sklearn.svm import SVR
from sklearn.neighbors import KNeighborsRegressor
```

```python
models = {
    "logitsRegression": LogisticRegression(),
    "KNN": KNeighborsRegressor(),
    "SVM": SVR(), 
    "decissionTree": DecisionTreeRegressor()
}
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
result = []

for name, model in models.items():
    model.fit(X_train_scaled, y_train)
    y_pred = model.predict(X_test_scaled)
    ME_error = mean_absolute_error(y_test, y_pred)
    MS_error = mean_squared_error(y_test, y_pred)
    R2_score = r2_score(y_test, y_pred)
    result.append({
        "model": name,
        "Mean_absolute_error": ME_error, 
        "Mean_scquared_error": MS_error,
        "R2_error": R2_score
    })
```

```python
result_df = pd.DataFrame(result)
result_df
```

```python
from sklearn.model_selection import KFold, cross_validate
from sklearn.pipeline import Pipeline
```

```python
kf = KFold(n_splits=4, shuffle=True, random_state=42)
```

```python
knn = Pipeline([
    ('scaler', StandardScaler()),
    ('model', KNeighborsRegressor(n_neighbors=5))
])

scores = cross_validate(
    knn,
    X_train,
    y_train,
    cv=kf,
    scoring=['r2', 'neg_mean_absolute_error', 'neg_root_mean_squared_error']
)
```

```python
print("R2 Scores:", scores['test_r2'])
print("MAE Scores:", -scores['test_neg_mean_absolute_error'])
print("RMSE Scores:", -scores['test_neg_root_mean_squared_error'])

print("\nAverage R2:", scores['test_r2'].mean())
print("Average MAE:", -scores['test_neg_mean_absolute_error'].mean())
print("Average RMSE:", -scores['test_neg_root_mean_squared_error'].mean())
```

```python
from sklearn.ensemble import GradientBoostingRegressor 
```

```python
GradientBoosting_pipeline = Pipeline([
    ('model', GradientBoostingRegressor(random_state=42))
])

GradientBoosting_param_grid = {
    'model__n_estimators': [100, 200, 300],
    'model__learning_rate': [0.03, 0.05, 0.1],
    'model__max_depth': [2, 3, 4],
    'model__min_samples_split': [2, 5, 10],
    'model__min_samples_leaf': [1, 2, 4],
    'model__subsample': [0.8, 1.0],
    'model__max_features': [None, 'sqrt']
}

GradientBoostingGrid = GridSearchCV(
    estimator=GradientBoosting_pipeline,
    param_grid=GradientBoosting_param_grid,
    cv=5,
    scoring='r2',
    n_jobs=-1,
    verbose=2
)

GradientBoostingGrid.fit(X_train, y_train)
```
